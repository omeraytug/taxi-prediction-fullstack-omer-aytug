# Preporcessing, Models, Evaluation, Imputation

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
from taxipred.utils.constants import TAXI_CSV_PATH


df = pd.read_csv(TAXI_CSV_PATH)
df.head()
```

```python
# Predict taxi price (target)
y = df["Trip_Price"]

# Use only non-leakage features
feature_cols = [
    "Trip_Distance_km",
    "Trip_Duration_Minutes",
    "Traffic_Conditions",
    "Weather",
    "Passenger_Count",
    "Time_of_Day",
    "Day_of_Week",
]

X = df[feature_cols]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.33, random_state=42)
```

```python
num_cols = ["Trip_Distance_km", "Trip_Duration_Minutes", "Passenger_Count"]
cat_cols = ["Traffic_Conditions", "Weather", "Time_of_Day", "Day_of_Week"]

numeric_transformer = SimpleImputer(strategy="median")

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, num_cols),
        ("cat", categorical_transformer, cat_cols)
    ]
)
```
