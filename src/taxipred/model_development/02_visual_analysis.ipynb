# Plots to Understand the Data

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from taxipred.utils.constants import TAXI_CSV_PATH


df = pd.read_csv(TAXI_CSV_PATH)
```

- Goal: Visualize relationships relevant to predicting Trip_Price.
- Note: We do not impute or drop columns here. This notebook is for visualization only.


```python
# This plot shows the distribution of taxi prices and helps identify skewness and outliers.
sns.histplot(df["Trip_Price"], kde=True)
plt.title("Distribution of Trip Price")
```

```python
# This plot shows how taxi price changes as trip distance increases.
sns.scatterplot(data=df, x="Trip_Distance_km", y="Trip_Price", alpha=0.4)
plt.title("Trip Price vs Trip Distance (km)")
```

```python
# This plot shows how taxi price changes as trip duration increases.
sns.scatterplot(data=df, x="Trip_Duration_Minutes", y="Trip_Price", alpha=0.4)
plt.title("Trip Price vs Trip Duration (minutes)")
```

```python
# This plot compares the distribution of taxi prices across traffic conditions.
sns.boxplot(data=df, x="Traffic_Conditions", y="Trip_Price")
plt.title("Trip Price by Traffic Conditions")
```

```python
# This plot compares the distribution of taxi prices across weather conditions.
sns.boxplot(data=df, x="Weather", y="Trip_Price")
plt.title("Trip Price by Weather Conditions")
```

```python
# This plot compares the distribution of taxi prices across different times of day.
sns.boxplot(data=df, x="Time_of_Day", y="Trip_Price")
plt.title("Trip Price by Time of Day")
```

```python
# This heatmap shows correlations between numerical variables and taxi price.
numeric_cols = [
    "Trip_Distance_km",
    "Trip_Duration_Minutes",
    "Passenger_Count",
    "Trip_Price"
]

sns.heatmap(df[numeric_cols].corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
```

```python
# This plot shows how taxi price varies with passenger count.
sns.boxplot(data=df, x="Passenger_Count", y="Trip_Price")
plt.title("Trip Price by Passenger Count")
```

```python
# This heatmap shows correlations between numerical features and the target (Trip_Price).
numeric_cols = df.select_dtypes(include="number").columns
corr = df[numeric_cols].corr()

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix (Numerical Features)")
```

### Key observations:
- Trip_Price increases with Trip_Distance_km.
- Trip_Price increases with Trip_Duration_Minutes.
- Traffic_Conditions and Time_of_Day shift the price distribution.
- Weather may affect prices depending on category differences.
